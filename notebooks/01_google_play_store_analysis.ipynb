```python
# Check the size column

print("Size data type:",apps["Size"].dtype)
print("Sample size values:")
print(apps["Size"].value_counts().head(15))
print("\nMissing values:",apps["Size"].isnull().sum())
```

```python
import numpy as np
import pandas as pd
def convert_size(size):
    if pd.isna(size):
        return np.nan
    size = str(size).strip()
    if size =="Values with device":
        return np.nan
    try:
        if size.endswith("M"):
            return float(size[:-1])
        elif size.endswith("k"):
            return float(size[:-1]) / 1024
        else: 
            return np.nan
    except ValueError:
        return np.nan

apps["Size_MB"]=apps["Size"].apply(convert_size)
print("New Size_MB column data type:",apps["Size_MB"].dtype)
display(apps[["App","Size","Size_MB"]].head(10))
```

```python
# Check whether duplicate review IDs belong to different apps

duplicate_reviews = reviews[
    reviews["reviewId"].duplicated(keep=False)
]

app_id_counts = (
    duplicate_reviews.groupby("reviewId")["appId"]
    .nunique()
)

print(
    "Duplicate IDs associated with multiple apps:",
    (app_id_counts > 1).sum()
)

print(
    "Duplicate IDs associated with one app:",
    (app_id_counts == 1).sum()
)
```

```python
# Check exact duplicates across all columns

print(
    "Exact duplicate rows:",
    reviews.duplicated().sum()
)

# Check duplicates across important review fields
review_key_columns = [
    "reviewId",
    "content",
    "score",
    "at",
    "appId"
]

print(
    "Duplicates across review fields:",
    reviews.duplicated(
        subset=review_key_columns
    ).sum()
)
```

```python
# Count review IDs that appear more than once

review_id_counts = reviews["reviewId"].value_counts()

repeated_review_ids = review_id_counts[
    review_id_counts > 1
]

print(
    "Number of distinct repeated review IDs:",
    len(repeated_review_ids)
)

print(
    "Total occurrences of repeated review IDs:",
    repeated_review_ids.sum()
)

print(
    "Extra occurrences beyond the first:",
    (repeated_review_ids - 1).sum()
)

print("\nFrequency of repeated IDs:")
print(repeated_review_ids.value_counts().sort_index())
```

```python
# Select the important review fields

review_key_columns = [
    "reviewId",
    "content",
    "score",
    "at",
    "appId"
]

# Display repeated records for investigation

duplicate_reviews = reviews[
    reviews["reviewId"].duplicated(keep=False)
].sort_values("reviewId")

display(
    duplicate_reviews[
        review_key_columns + [
            "userName",
            "thumbsUpCount"
        ]
    ].head(20)
)
```

```python
# Keep the first occurrence of each review ID

reviews_clean = reviews.drop_duplicates(
    subset=["reviewId"],
    keep="first"
).copy()

print("Original dataset shape:", reviews.shape)
print("Cleaned dataset shape:", reviews_clean.shape)

print(
    "Remaining duplicate review IDs:",
    reviews_clean["reviewId"].duplicated().sum()
)
```

```python
# Check missing values in the cleaned reviews dataset

print("Cleaned dataset shape:", reviews_clean.shape)

print("\nMissing values in each column:")
display(
    reviews_clean.isnull()
    .sum()
    .to_frame(name="Missing Values")
)

# Calculate missing value percentages
missing_percentage = (
    reviews_clean.isnull().mean() * 100
).round(2)

print("\nMissing value percentages:")
display(
    missing_percentage[missing_percentage > 0]
    .to_frame(name="Missing Percentage (%)")
)
```

```python
# Check empty review text

empty_reviews = (
    reviews_clean["content"]
    .fillna("")
    .str.strip()
    .eq("")
    .sum()
)

print("Empty review texts:", empty_reviews)

print(
    "Missing review texts:",
    reviews_clean["content"].isnull().sum()
)
```

```python
# Convert review dates to datetime

reviews_clean["at"] = pd.to_datetime(
    reviews_clean["at"],
    errors="coerce",
    utc=True
)

print("Review date data type:", reviews_clean["at"].dtype)

print("\nSample review dates:")
display(
    reviews_clean[["content", "score", "at"]].head(10)
)

print(
    "\nMissing review dates:",
    reviews_clean["at"].isnull().sum()
)
```

```python
# Final verification of the cleaned reviews dataset

print("Final dataset shape:", reviews_clean.shape)

print(
    "\nDuplicate review IDs:",
    reviews_clean["reviewId"].duplicated().sum()
)

print(
    "Exact duplicate rows:",
    reviews_clean.duplicated().sum()
)

print(
    "Missing review text:",
    reviews_clean["content"].isnull().sum()
)

print(
    "Empty review text:",
    reviews_clean["content"]
    .fillna("")
    .str.strip()
    .eq("")
    .sum()
)

print(
    "Invalid review scores:",
    (~reviews_clean["score"].between(1, 5)).sum()
)

print(
    "Missing review dates:",
    reviews_clean["at"].isnull().sum()
)
```

```python
# Save the cleaned reviews dataset

reviews_clean.to_csv(
    "../data/reviews_clean.csv",
    index=False
)

print("Cleaned reviews dataset saved successfully!")
```

## 3. Data Cleaning and Preparation

In this section, we will investigate duplicate records,
correct data types, and prepare columns for exploratory
data analysis.

```python
# Investigate duplicate app names

duplicate_app_names = apps[
    apps["App"].duplicated(keep=False)
].sort_values("App")

display(
    duplicate_app_names[
        ["App", "Category", "Rating", "Reviews", "Installs", "Price"]
    ].head(20)
)
```

```python
# Create a working copy

apps_eda = apps.copy()

# Remove exact duplicate rows

apps_eda = apps_eda.drop_duplicates().copy()

print("Original dataset shape:", apps.shape)
print("After removing exact duplicates:", apps_eda.shape)
print("Remaining exact duplicates:", apps_eda.duplicated().sum())
```

```python
# Check remaining duplicate app names

print(
    "Repeated app names after exact duplicate removal:",
    apps_eda["App"].duplicated().sum()
)

# Display examples for further investigation

remaining_duplicates = apps_eda[
    apps_eda["App"].duplicated(keep=False)
].sort_values("App")

display(
    remaining_duplicates[
        ["App", "Category", "Rating", "Reviews", "Installs", "Price"]
    ].head(20)
)
```

```python
# Convert Reviews to numeric

apps_eda["Reviews"] = pd.to_numeric(
    apps_eda["Reviews"],
    errors="coerce"
)

# Clean and convert Installs

apps_eda["Installs"] = (
    apps_eda["Installs"]
    .astype("string")
    .str.replace(",", "", regex=False)
    .str.replace("+", "", regex=False)
)

apps_eda["Installs"] = pd.to_numeric(
    apps_eda["Installs"],
    errors="coerce"
)

# Clean and convert Price

apps_eda["Price"] = (
    apps_eda["Price"]
    .astype("string")
    .str.replace("$", "", regex=False)
)

apps_eda["Price"] = pd.to_numeric(
    apps_eda["Price"],
    errors="coerce"
)

# Check converted data types

print("Updated data types:")

display(
    apps_eda[["Reviews", "Installs", "Price"]]
    .dtypes
    .to_frame(name="Data Type")
)

print("\nSample converted values:")

display(
    apps_eda[["App", "Reviews", "Installs", "Price"]].head(10)
)
```

```python
# Convert app size to MB

def convert_size_to_mb(size):
    if pd.isna(size):
        return np.nan

    size = str(size).strip()

    if size == "Varies with device":
        return np.nan

    try:
        if size.endswith("M"):
            return float(size[:-1])

        elif size.endswith("k"):
            return float(size[:-1]) / 1024

    except ValueError:
        return np.nan

    return np.nan


apps_eda["Size_MB"] = apps_eda["Size"].apply(convert_size_to_mb)

print("Missing Size_MB values:", apps_eda["Size_MB"].isnull().sum())

display(apps_eda[["App", "Size", "Size_MB"]].head(10))
```

```python
import pandas as pd
import numpy as np

# Function to convert app size into MB
def convert_size(size):

    if pd.isna(size):
        return np.nan

    size = str(size).strip()

    if size == "Varies with device":
        return np.nan

    try:
        if size.endswith("M"):
            return float(size[:-1])

        elif size.endswith("k"):
            return float(size[:-1]) / 1024

        else:
            return float(size)

    except ValueError:
        return np.nan


# Create a new numeric column
apps_eda["Size_MB"] = apps_eda["Size"].apply(convert_size)

# Check results
print(apps_eda[["Size", "Size_MB"]].head(10))
```

### Sentiment Analysis of User Reviews

This section analyzes user reviews to understand customer
sentiment toward applications. Reviews will be classified
as Positive, Neutral, or Negative using their review scores.

The analysis will help identify customer satisfaction levels
and opportunities for app improvement.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned reviews dataset
reviews_df = pd.read_csv("../data/reviews_clean.csv")

# Keep only rows with review text and valid scores
sentiment_df = reviews_df[["content", "score"]].copy()

sentiment_df["content"] = sentiment_df["content"].astype("string").str.strip()
sentiment_df["score"] = pd.to_numeric(
    sentiment_df["score"], errors="coerce"
)

sentiment_df = sentiment_df.dropna(subset=["content", "score"])

sentiment_df = sentiment_df[
    (sentiment_df["content"] != "") &
    (sentiment_df["score"].between(1, 5))
]

# Remove duplicate reviews
sentiment_df = sentiment_df.drop_duplicates(
    subset=["content", "score"]
)

print("Valid reviews:", len(sentiment_df))
print("\nReview score distribution:")
print(sentiment_df["score"].value_counts().sort_index())
```

```python
def classify_sentiment(score):
    if score >= 4:
        return "Positive"
    elif score == 3:
        return "Neutral"
    else:
        return "Negative"


sentiment_df["Sentiment"] = sentiment_df["score"].apply(
    classify_sentiment
)

print("Sentiment classification completed.")

print("\nSentiment counts:")
print(sentiment_df["Sentiment"].value_counts())

print("\nSentiment percentages:")
print(
    (sentiment_df["Sentiment"].value_counts(normalize=True) * 100)
    .round(2)
)
```

```python
sentiment_counts = sentiment_df["Sentiment"].value_counts()

plt.figure(figsize=(9, 5))

sentiment_colors ={ 
    "Negative": "#E15759",
    "Neutral": "#EDC948",
    "Positive": "#59A14F"
}

ax = sns.barplot(
    x=sentiment_counts.index,
    y=sentiment_counts.values,
    hue= sentiment_counts.index,
    palette=sentiment_colors,
    legend=False
)

plt.title("Distribution of User Review Sentiment", fontsize=14, fontweight = "bold")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")

for container in ax.containers:
    ax.bar_label(container, fmt="%.0f", padding=3)

plt.tight_layout()
plt.savefig(
    "../images/review_sentiment_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

#### Business Insights

- Negative reviews account for approximately 33.86% of the analyzed reviews.
- Neutral reviews represent approximately 33.35%.
- Positive reviews account for approximately 32.74%.
- The distribution indicates that customer experiences are mixed.
- Further analysis of review text is required to identify the
  specific reasons behind negative customer experiences.

### Review Text Analysis: Common Customer Feedback

This section examines user review text to identify frequently
mentioned words and recurring customer concerns.

The analysis aims to discover common themes in customer feedback
that may help developers improve application usability,
performance, and customer satisfaction.

```python
import pandas as pd
import re
from collections import Counter

# Load cleaned reviews
reviews_df = pd.read_csv("../data/reviews_clean.csv")

# Remove missing and empty reviews
text_df = reviews_df.dropna(subset=["content"]).copy()

text_df["content"] = text_df["content"].astype(str).str.strip()

text_df = text_df[text_df["content"] != ""]

print("Total valid text reviews:", len(text_df))
```

```python
import re
from collections import Counter
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Portuguese stopwords
stopwords = {
    "que", "com", "muito", "para", "mais", "mas", "tem",
    "uma", "por", "bom", "pra", "quando", "não", "nao",
    "dos", "das", "nos", "nas", "uma", "uns", "umas",
    "de", "do", "da", "em", "no", "na", "os", "as",
    "e", "ou", "se", "eu", "me", "minha", "meu",
    "seu", "sua", "ele", "ela", "eles", "elas",
    "foi", "ser", "ter", "está", "esta", "estou",
    "são", "sao", "isso", "esse", "essa", "aqui",
    "já", "ja", "ao", "aos", "pela", "pelo",
    "como", "bem", "app", "aplicativo"
}

# Extract Portuguese words from review text
all_words = []

for review in text_df["content"]:
    words = re.findall(
        r"\b[a-zA-ZÀ-ÿ]{3,}\b",
        str(review).lower()
    )

    all_words.extend(
        word for word in words
        if word not in stopwords
    )

# Count word frequencies
word_counts = Counter(all_words)

# Display the 15 most frequent words
top_words = word_counts.most_common(15)

word_freq_df = pd.DataFrame(
    top_words,
    columns=["Word", "Frequency"]
)

display(word_freq_df)
```

```python
plt.figure(figsize=(12, 7))

ax = sns.barplot(
    data=word_freq_df,
    x="Frequency",
    y="Word"
)

plt.title(
    "Top 15 Most Frequent Words in User Reviews",
    fontsize=15
)

plt.xlabel("Word Frequency")
plt.ylabel("Words")

for container in ax.containers:
    ax.bar_label(container, fmt="%.0f", padding=3)

plt.tight_layout()

plt.savefig(
    "../images/top_15_review_words.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

###  Analysis of Negative Customer Reviews

This section examines low-rated reviews to identify recurring
customer concerns related to orders, delivery, payments,
coupons, and other app experiences.

The objective is to identify potential improvement areas
using actual customer feedback.

```python
# Select review text and rating
complaints_df = reviews_df[
    ["content", "score"]
].copy()

# Convert ratings to numeric
complaints_df["score"] = pd.to_numeric(
    complaints_df["score"],
    errors="coerce"
)

# Remove missing and invalid records
complaints_df = complaints_df.dropna(
    subset=["content", "score"]
)

complaints_df["content"] = (
    complaints_df["content"].astype(str).str.strip()
)

complaints_df = complaints_df[
    complaints_df["content"].ne("") &
    complaints_df["score"].between(1, 5)
]

# Select low-rated reviews
negative_reviews = complaints_df[
    complaints_df["score"] <= 2
].copy()

print("Total valid reviews:", len(complaints_df))
print("Negative reviews:", len(negative_reviews))

display(negative_reviews.head(10))
```

```python
# Define business-related topics
topics = {
    "Orders": ["pedido", "pedidos"],
    "Delivery": ["entrega", "entregas", "entregador"],
    "Payments": ["pagamento", "pagamentos", "cartão", "cartao"],
    "Coupons": ["cupom", "cupons", "desconto", "descontos"],
    "Restaurants": ["restaurante", "restaurantes"],
    "Problems": ["problema", "problemas", "erro", "falha"]
}

# Count negative reviews mentioning each topic
topic_counts = {}

for topic, keywords in topics.items():
    pattern = (
        r"\b(?:"
        + "|".join(re.escape(word) for word in keywords)
        + r")\b"
    )

    topic_counts[topic] = (
        negative_reviews["content"]
        .str.contains(pattern, case=False, regex=True, na=False)
        .sum()
    )

topic_df = pd.DataFrame(
    list(topic_counts.items()),
    columns=["Topic", "Negative Reviews"]
).sort_values(
    "Negative Reviews",
    ascending=False
)

display(topic_df)
```

```python
plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=topic_df,
    x="Negative Reviews",
    y="Topic"
)

plt.title("Common Topics in Negative User Reviews")
plt.xlabel("Number of Negative Reviews Mentioning Topic")
plt.ylabel("Topic")

for container in ax.containers:
    ax.bar_label(container, fmt="%.0f", padding=3)

plt.tight_layout()

plt.savefig(
    "../images/negative_review_topics.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
